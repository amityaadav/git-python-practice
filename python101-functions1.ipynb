This is an example of a basic function.  The function evaluates whether a number is even.

```python
def even_check(number):
    result = number % 2 == 0
    return result
```

```python
even_check(5)
```

```python
even_check(4)
```

```python
even_check(4.4)
```

```python
even_check(4.0)
```

If your function is very straight forward, you can have the function and the return in the same line as below

```python
def even_check2(number2):
    return number2 % 2 == 0
```

```python
even_check2(5)
```

```python
even_check2(4)
```
