In this function, we will use a list to loop through even/odd evaluation

```python
def check_even_list(myList):  
    for number in myList:
        if number % 2 == 0:
            return True
        else:
            pass  # pass without doing anything if the list has odd number
```

In the following call, nothing is returned because there were no even numbers, and therefore the function "pass"es through the results

```python
check_even_list([1,3,5])
```

The following call returns True because it matches the condition in the function

```python
check_even_list([2,6,8])
```

Now two edge cases, 1st with an even no. first and then 2nd with an odd number first

```python
check_even_list([1,2,3,4])
```

```python
check_even_list([2,3,4,5])
```

When the number is odd, you want to return false, but where to add the return condition matters a lot in the output.  

1st: if you add the "return", indented within the if condition, the evaluation loop is only run for 1 iteration.

```python
def check_even_list(myList):  
    for number in myList:
        if number % 2 == 0:
            return True
        else:
            return False
```

```python
check_even_list([1,2])
```

2nd: if you add the "return" keyword indented to the for loop, it will go through every item in the list

```python
def check_even_list(myList):  
    for number in myList:
        if number % 2 == 0:
            return True
        else:
            pass
    return False
```

```python
check_even_list([1,3,4]). # notice that no. 4 is at the end of list item, therefore it should return true if it's going through every item
```

```python
check_even_list([1,5,3]) # this list only has odd numbers
```
