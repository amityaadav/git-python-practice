This function will print all even numbers from a list

```python
def check_even_list(myList):
    even_numbers = []

    for number in myList:
        if number % 2 == 0:
            even_numbers.append(number)
        else:
            pass
    return even_numbers
```

```python
check_even_list([1,3,5])
```

```python
check_even_list([1,3,4,5])
```

```python
check_even_list([1,7,6,4])
```
