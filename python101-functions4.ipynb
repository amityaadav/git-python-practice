Unpacking tuples through functions

```python
stock_prices = [('Apple',100),('NVDA',200),('MSFT',300)] # A sample tuple
```

Now to unpack this, you can either unpack it by using assigning key in the loop or print everything by assigning 1 variable to the list

```python
# Assign 1 var to the list to print everything
for item in stock_prices:
    print(item)
```

```python
# Assign var to each tuple item
for ticker,price in stock_prices:
    print(ticker)
```

```python
# or modify existing data
for ticker,price in stock_prices:
    print(price * 1.2)
```
