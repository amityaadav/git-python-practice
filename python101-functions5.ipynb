```python
employee_work_hours = [('Amit',40),('Becky',60),('Raj',30)]
```

```python
def employee_check(employee_work_hours):

    current_max = 0                     # initialize as 0
    employee_of_month = ''              # initialize as empty

    for employee,hours in employee_work_hours:
        if hours > current_max:
            current_max = hours
            employee_of_month = employee
        else:
            pass

```

```python
employee_check(employee_work_hours)
```
